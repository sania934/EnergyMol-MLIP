```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski
import pandas as pd

records = []

for mol_info in molecules:

    mol = Chem.MolFromSmiles(mol_info["smiles"])

    if mol is None:
        raise ValueError(
            f"Invalid SMILES for {mol_info['id']}: "
            f"{mol_info['smiles']}"
        )

    canonical_smiles = Chem.MolToSmiles(
        mol,
        canonical=True
    )

    formula = Chem.rdMolDescriptors.CalcMolFormula(mol)

    records.append({
        "id": mol_info["id"],
        "name": mol_info["name"],
        "family": mol_info["family"],
        "input_smiles": mol_info["smiles"],
        "canonical_smiles": canonical_smiles,
        "formula": formula,
        "molecular_weight": Descriptors.MolWt(mol),
        "total_atoms": mol.GetNumAtoms(),
        "heavy_atoms": mol.GetNumHeavyAtoms(),
        "carbon": sum(
            1 for atom in mol.GetAtoms()
            if atom.GetSymbol() == "C"
        ),
        "hydrogen": sum(
            1 for atom in mol.GetAtoms()
            if atom.GetSymbol() == "H"
        ),
        "nitrogen": sum(
            1 for atom in mol.GetAtoms()
            if atom.GetSymbol() == "N"
        ),
        "oxygen": sum(
            1 for atom in mol.GetAtoms()
            if atom.GetSymbol() == "O"
        ),
        "sulfur": sum(
            1 for atom in mol.GetAtoms()
            if atom.GetSymbol() == "S"
        ),
        "fluorine": sum(
            1 for atom in mol.GetAtoms()
            if atom.GetSymbol() == "F"
        ),
        "rings": Lipinski.RingCount(mol),
        "aromatic_rings": Lipinski.NumAromaticRings(mol),
        "rotatable_bonds": Lipinski.NumRotatableBonds(mol),
    })

df = pd.DataFrame(records)

print("All SMILES successfully parsed.")
display(df)
```

```python
duplicates = df[
    df["canonical_smiles"].duplicated(keep=False)
].sort_values("canonical_smiles")

if len(duplicates) == 0:
    print("PASS: No duplicate molecular structures detected.")
else:
    print("WARNING: Duplicate structures detected.")
    display(duplicates)
```

```python
element_columns = [
    "carbon",
    "hydrogen",
    "nitrogen",
    "oxygen",
    "sulfur",
    "fluorine"
]

element_totals = df[element_columns].sum()

print("Elemental coverage:")
display(element_totals.to_frame("total_count"))
```

```python
family_counts = (
    df["family"]
    .value_counts()
    .sort_index()
)

display(family_counts.to_frame("number_of_molecules"))
```
